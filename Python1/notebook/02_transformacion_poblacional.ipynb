```python
import pandas as pd
import requests

# Fetch the data.
df = pd.read_csv("https://ourworldindata.org/grapher/life-expectancy.csv?v=1&csvType=full&useColumnShortNames=true", storage_options = {'User-Agent': 'Our World In Data data fetch/1.0'})

# Fetch the metadata
metadata = requests.get("https://ourworldindata.org/grapher/life-expectancy.metadata.json?v=1&csvType=full&useColumnShortNames=true").json()
```

```python
df.head(3)
```

```python
#Que buscque la cadenas de texto que comiencen con tal letra
#Sirve para cambiarle el nombre a la columna "entity" y que ahora se llame "Pais" en tu tabla de datos.
df.rename(columns={"entity": "Pais"}, inplace=True)
```

```python
df.head()
```

```python
df.set_index("Pais")
```

```python
#Sirve para convertir la columna "Pais" en el identificador principal (índice) de las filas de tu tabla
df.set_index(["code","year"])
```

```python
#Te lanza todos los nulos que hay en code
df.loc[df["code"].isnull()]
```

```python
#RECARGUE EL DATASET porque lo converti en un indeice de filas entity 
df.loc[df["code"].notnull(),["entity","year","life_expectancy_0"]].head(3)
```

```python
#Para borrar los valores nulos que hay en code
df.dropna(subset=["code"], inplace=True)
df.isnull().sum()
```

```python
df.shape
```

```python
#Calcular el promedio de esperanza de vida por cada país o entidad (entity), ordenarlos de menor a mayor y quedarse con los valores más bajos.
resultado = (
    df.groupby("entity")["life_expectancy_0"]
    .mean()
    .sort_values(ascending=True)
)
resultado.head(10)
```

```python
#
promedio_pais = (
    df.groupby("entity")["life_expectancy_0"]
    .mean()
    .round(2)
    .sort_values(ascending=False)
)

print("Promediio de esperanza de vida por pais", promedio_pais)
```
