```python
df_historico = df.loc[df["Year"] == 2020]

col = df_historico["Annual change in population"]

mean = col.mean()
print("1.	¿Cuál es el valor de la media?", mean)

mediana = col.median()
print("2.	¿Cuál es el valor de la mediana?", mediana)

desviacion_estandar = col.std()
print("3.	¿Cuál es el valor de la desviación estándar?", desviacion_estandar)

minimo = col.min()
maximo = col.max()
print("4.	¿Qué entidad presenta el valor mínimo y cuál el máximo? Maximo: ", maximo,("Minimo: "), minimo)
```
